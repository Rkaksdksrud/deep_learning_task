## 신규 비교의 판단

- 회전 단독은 기존 반전+회전 대비 Δ−0.26/−2.69/−1.37%p, 평균−1.44%p로 세 조건 모두 하락했다. 현재 정책에서 반전을 제거한 비교이며 flip-only와의 비교는 아니다.
- 밝기는 기존 반전+회전에 factor U[.85,1.15]를 추가했다. Δ+1.34/−2.10/−0.30%p, 평균−0.35%p로 첫 조건의 개선이 유지되지 않았다. 단발 최고91.81%를 새 기준으로 채택하지 않았다.
- 대비는 같은 기하 정책에 contrast만 별도 추가했다. 첫 조건91.46%, Δ+0.98927%p로 사전 추가 확인 기준 +1%p에 못 미쳐 나머지2조건을 실행하지 않았다.
- 낮춘 LR의 이번 무증강 추가 seed는60.11→63.94%다. 세 조건 평균 개선이 있어도 특정 환자의 악화가 남으며 기존 geometry 모델보다 좋은 설정이라는 뜻은 아니다.

3조건 평균 변화의 표본 SD는 회전1.22%p·밝기1.72%p·LR2.53%p다. 이는 조건 간 기술 통계이며 신뢰구간이 아니다. 밝기/대비 중 첫 조건 상승 기준으로 하나만 추가 확인해 후보별 예산과 선택 편향이 남는다. 새8run 모두 train BA 관측 최대는100%지만 validation과 격차가 남았으며, 과적합과 환자 분포 차이의 원인을 분리한 증거는 아니다.

## 공개 manifest에서 전체 모델 목록 읽기

다음 셀은 공개된 네 manifest의 ID·점수·다운로드 URL만 읽는다. 파일을 다운로드하거나 신경망을 실행하지 않는다. GitHub 렌더링의 출력은 비워 두었으며, 아래 표는 각자가 실행해서 확인한다.

````python
from pathlib import Path
import json
import pandas as pd

candidates = [Path.cwd(), *Path.cwd().parents]
repo = next(p for p in candidates if (p / "models/task2-resnet18-20261111/manifest.json").is_file())
bundles = ["task2-resnet18-20261111", "task2-followups-20261118", "task2-training-settings-20261124", "task2-augmentation-lr-20261125", "task2-legacy-baseline-20261101"]
rows = []
for bundle in bundles:
    manifest = json.loads((repo / "models" / bundle / "manifest.json").read_text(encoding="utf-8"))
    for model in manifest["models"]:
        rows.append({"bundle": bundle, "run_id": model["id"], "fold": model["validation_fold"], "seed": model["seed"], "BA_percent": 100 * model["balanced_accuracy"], "download_url": model["download_url"]})
catalog = pd.DataFrame(rows)
assert len(catalog) == 29
catalog
````
